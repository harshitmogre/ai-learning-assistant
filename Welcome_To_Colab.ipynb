```
from sklearn.impute import SimpleImputer

mean_imputer = SimpleImputer(strategy="mean")

employee_mean = employee.copy()

numerical_cols = employee.select_dtypes(include="number").columns

employee_mean[numerical_cols] = mean_imputer.fit_transform(
    employee[numerical_cols]
)
employee_mean
```

```
median_imputer = SimpleImputer(strategy="median")

employee_median = employee.copy()

employee_median[numerical_cols] = median_imputer.fit_transform(
    employee[numerical_cols]
)

print(employee_median);
```

```
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=3)

employee_knn = employee.copy()

employee_knn[numerical_cols] = knn_imputer.fit_transform(
    employee[numerical_cols]
)
print(employee_knn);
```

```
# Method 1: Boxplot
import seaborn as sns
import matplotlib.pyplot as plt

columns = [
    "Age",
    "Experience",
    "Monthly_Hours",
    "Salary"
]

for col in columns:

    plt.figure(figsize=(5,4))

    sns.boxplot(y=employee[col])

    plt.title(col)

    plt.show()
```

```
# Method 2: IQR Method
Q1 = employee["Salary"].quantile(0.25)

Q3 = employee["Salary"].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers = employee[
    (employee["Salary"]<lower) |
    (employee["Salary"]>upper)
]

outliers
```

```
# Method 3: Z-Score
from scipy.stats import zscore

employee["Salary_Z"] = zscore(employee["Salary"])

outliers = employee[
    abs(employee["Salary_Z"]) > 3
]
```

```
# Handling Outliers

# Removal
employee_removed = employee[
    (employee["Salary"]>=lower) &
    (employee["Salary"]<=upper)
]
employee_removed
```

```
# Capping
employee_cap = employee.copy()

employee_cap["Salary"] = employee_cap["Salary"].clip(
    lower,
    upper
)
employee_cap
```

```
# Feature Scaling
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

employee_minmax = employee.copy()

employee_minmax[numerical_cols] = scaler.fit_transform(
    employee[numerical_cols]
)

# Code

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

employee_standard = employee.copy()

employee_standard[numerical_cols] = scaler.fit_transform(
    employee[numerical_cols]
)
```

```
# Compare Original vs Scaled
comparison = pd.DataFrame({
    "Original": employee["Salary"],
    "MinMax": employee_minmax["Salary"],
    "Standard": employee_standard["Salary"]
})

comparison.head()
```

```
# Histogram Comparison

plt.figure(figsize=(8,5))

plt.hist(employee["Salary"], alpha=0.5, label="Original")
plt.hist(employee_minmax["Salary"], alpha=0.5, label="MinMax")
plt.hist(employee_standard["Salary"], alpha=0.5, label="Standard")

plt.legend()
plt.title("Salary Distribution")
plt.show()
```

```
# Boxplot Comparison

plot_df = pd.DataFrame({
    "Original": employee["Salary"],
    "MinMax": employee_minmax["Salary"],
    "Standard": employee_standard["Salary"]
})

plt.figure(figsize=(8,5))
sns.boxplot(data=plot_df)
plt.title("Comparison of Scaling Methods")
plt.show()
```
